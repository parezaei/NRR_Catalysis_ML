```python
#Done
def vasp_to_coordinate(coord_line, cell_matrix, atom_type, fractional =False):
    coord_list = coord_line.split()[0:3]
    #atom_type = coord_line.split()[0]
    for n in range(len(coord_list)):
        #coord_list[n] = re.sub("d","0",coord_list[n])
        coord_list[n] = float(coord_list[n])
    coord_array = np.array(coord_list).reshape(1,3)
    if fractional == False:
        coord_array_ang = np.dot(coord_array,cell_matrix)
        return (atom_type,coord_array_ang) 
    else:
        return (atom_type,coord_array)
```

```python
import numpy as np
import pandas as pd
from pymatgen.core import Structure

def read_CONTCAR(file_path, fractional=False, output_type='df'):
    """
    Reads a CONTCAR file and returns atomic coordinates as a list or pandas DataFrame.

    Parameters:
        file_path (str): Path to the CONTCAR file.
        fractional (bool): If True, return fractional coordinates. If False, return real coordinates.
        output_type (str): 'list' for a list of coordinates (as 3×1 NumPy arrays), 'df' for a pandas DataFrame.

    Returns:
        list or pd.DataFrame: Atomic coordinates in requested format.
    """
    # Load structure using pymatgen
    structure = Structure.from_file(file_path)
    
    # Get atomic symbols
    atom_types = [site.species_string for site in structure]
    
    # Get coordinates (fractional or real)
    coordinates = structure.frac_coords if fractional else structure.cart_coords

    # If output_type is 'list', return a list with coordinates as (atom_type, 3×1 NumPy array)
    if output_type == 'list':
        return [(atom, np.array([[x], [y], [z]])) for atom, (x, y, z) in zip(atom_types, coordinates)]
    
    # If output_type is 'df', return a pandas DataFrame
    elif output_type == 'df':
        df = pd.DataFrame(coordinates, columns=['x', 'y', 'z'])
        df.insert(0, 'atom_type', atom_types)
        return df
    
    else:
        raise ValueError("Invalid output_type. Choose 'list' or 'df'.")

```

```python
#Done
def calculate_nearest_slab_distances_with_atom_info_old(poscar_file, molecule_atom_types, num_neighbor = 9):
    """
    Calculates the distances of the 9 nearest atoms in the slab from the molecule center.
    Includes the atom type, Cartesian coordinates, distance vector, distance magnitude, 
    and the order of each atom in the POSCAR file.

    Args:
        poscar_file (str): Path to the POSCAR file.
        molecule_atom_types (list): List of molecule atom types to identify (e.g., ['N', 'H']).
        num_neighbor (int): Number of nearest neighbors to return. Default is 9.

    Returns:
        dict: A dictionary containing:
            - 'molecule_center': Cartesian coordinates of the molecule center.
            - 'nearest_atoms': A list of the 9 nearest atoms with the following info:
                - Atom type (e.g., 'Pt', 'N').
                - Cartesian coordinates as a tuple.
                - Distance vector (relative position from molecule center).
                - Distance from the molecule center (magnitude of distance vector).
                - Atom order in the POSCAR file (0-based index).
    """
    import numpy as np
    from pymatgen.io.vasp import Poscar

    # Load POSCAR file and extract structure
    with open(poscar_file, "r") as f:
        file_content = f.read()  # Read file contents to ensure it's accessible

    poscar = Poscar.from_string(file_content)
    structure = poscar.structure

    # Convert fractional coordinates to Cartesian coordinates
    lattice_matrix = structure.lattice.matrix  # Lattice vectors (3x3 matrix)
    fractional_coords = structure.frac_coords  # Fractional coordinates of all atoms
    all_coords = np.dot(fractional_coords, lattice_matrix)  # Convert fractional to Cartesian
    all_atom_types = structure.species  # List of atom types

    # Filter molecule atoms by type
    molecule_coords = []
    molecule_indices = []
    for i, atom_type in enumerate(all_atom_types):
        if str(atom_type) in molecule_atom_types:
            molecule_coords.append(all_coords[i])
            molecule_indices.append(i)

    if len(molecule_coords) == 0:
        raise ValueError(f"No atoms of the specified types {molecule_atom_types} were found in the POSCAR file.")

    molecule_coords = np.array(molecule_coords)

    # Explicitly extract coordinates of N atoms
    nitrogen_coords = [
        all_coords[i] for i, atom in enumerate(all_atom_types) if str(atom) == "N"
    ]

    # Determine the molecule center based on the specified strategy
    if len(nitrogen_coords) == 2:
        # Two N atoms: Use the center of the two N atoms
        molecule_center = np.mean(nitrogen_coords, axis=0)
    elif len(nitrogen_coords) == 1:
        # One N atom: Use its coordinate
        molecule_center = nitrogen_coords[0]
    elif len(molecule_coords) == 2:
        # Biatomic molecule (not N₂): Use the center of the two atoms
        molecule_center = np.mean(molecule_coords, axis=0)
    else:
        # No N atoms or other multi-atom molecule: Use the center of all atoms
        molecule_center = np.mean(molecule_coords, axis=0)

    # Calculate distances from the molecule center to all slab atoms
    distances = []
    slab_atoms_info = []
    for i, coord in enumerate(all_coords):
        # Exclude distances between molecule atoms themselves
        if i in molecule_indices:
            continue
        distance_vector = coord - molecule_center  # Compute the distance vector
        distance = np.linalg.norm(distance_vector)  # Compute the magnitude of the distance vector
        distances.append(distance)
        slab_atoms_info.append((str(all_atom_types[i]), tuple(coord), tuple(distance_vector), distance, i))  # 0-based index

    # Sort by distance and select the nearest num_neighbor atoms
    sorted_atoms_info = sorted(slab_atoms_info, key=lambda x: x[3])[:num_neighbor]

    # Prepare results
    result = {
        "molecule_center": molecule_center.tolist(),
        "nearest_atoms": sorted_atoms_info  # Each entry contains (type, coordinates, distance vector, distance magnitude, index)
    }
    return result
```

```python
def calculate_nearest_slab_distances_with_atom_info(poscar_file, molecule_atom_types, num_neighbor=9):
    """
    Calculates the distances of the 9 nearest atoms in the slab from the molecule center.
    Includes the atom type, Cartesian coordinates, distance vector, distance magnitude, 
    and the order of each atom in the POSCAR file.

    Args:
        poscar_file (str): Path to the POSCAR file.
        molecule_atom_types (list): List of molecule atom types to identify (e.g., ['N', 'H']).
        num_neighbor (int): Number of nearest neighbors to return. Default is 9.

    Returns:
        dict: A dictionary containing:
            - 'molecule_center': Cartesian coordinates of the molecule center.
            - 'nearest_atoms': A list of the 9 nearest atoms with the following info:
                - Atom type (e.g., 'Pt', 'N').
                - Cartesian coordinates as a tuple.
                - Distance vector (relative position from molecule center).
                - Distance from the molecule center (magnitude of distance vector).
                - Atom order in the POSCAR file (0-based index).
            - 'mol_coords': A list of tuples (atom_type, coordinate)
    """
    import numpy as np
    from pymatgen.io.vasp import Poscar

    # Load POSCAR file and extract structure
    with open(poscar_file, "r") as f:
        file_content = f.read()

    poscar = Poscar.from_string(file_content)
    structure = poscar.structure

    # Convert fractional coordinates to Cartesian coordinates
    lattice_matrix = structure.lattice.matrix
    fractional_coords = structure.frac_coords
    all_coords = np.dot(fractional_coords, lattice_matrix)
    all_atom_types = structure.species

    # Filter molecule atoms by type
    molecule_coords = []
    molecule_indices = []
    molecule_types = []
    for i, atom_type in enumerate(all_atom_types):
        if str(atom_type) in molecule_atom_types:
            molecule_coords.append(all_coords[i])
            molecule_indices.append(i)
            molecule_types.append(str(atom_type))

    if len(molecule_coords) == 0:
        raise ValueError(f"No atoms of the specified types {molecule_atom_types} were found in the POSCAR file.")

    molecule_coords = np.array(molecule_coords)

    # Explicitly extract coordinates of N atoms
    nitrogen_coords = [
        all_coords[i] for i, atom in enumerate(all_atom_types) if str(atom) == "N"
    ]

    # Determine the molecule center based on the specified strategy
    if len(nitrogen_coords) == 2:
        molecule_center = np.mean(nitrogen_coords, axis=0)
    elif len(nitrogen_coords) == 1:
        molecule_center = nitrogen_coords[0]
    elif len(molecule_coords) == 2:
        molecule_center = np.mean(molecule_coords, axis=0)
    else:
        molecule_center = np.mean(molecule_coords, axis=0)

    # Calculate distances from the molecule center to all slab atoms
    distances = []
    slab_atoms_info = []
    for i, coord in enumerate(all_coords):
        if i in molecule_indices:
            continue
        distance_vector = coord - molecule_center
        distance = np.linalg.norm(distance_vector)
        slab_atoms_info.append((str(all_atom_types[i]), tuple(coord), tuple(distance_vector), distance, i))

    sorted_atoms_info = sorted(slab_atoms_info, key=lambda x: x[3])[:num_neighbor]

    # ✅ NEW: Store molecule atom types and coordinates as a list of tuples
    mol_coords = [(mol_type, tuple(coord)) for mol_type, coord in zip(molecule_types, molecule_coords)]

    result = {
        "molecule_center": molecule_center.tolist(),
        "nearest_atoms": sorted_atoms_info,
        "mol_coords": mol_coords
    }
    return result
```

```python
from pymatgen.io.vasp import Poscar
from pymatgen.core.structure import Structure

def calculate_molecule_distances(poscar_file, molecule_atoms):
    """
    Reads a CONTCAR file, extracts a molecular structure, and calculates distances between the molecule's atoms,
    considering periodic boundary conditions (PBC).

    Args:
        poscar_file (str): Path to the CONTCAR file.
        molecule_atoms (list): List of atomic symbols representing the molecule (e.g., ["N", "H"] for NH3).

    Returns:
        list: A sorted list of calculated distances based on the molecule type.
    """
    try:
        # Read CONTCAR file as a string
        with open(poscar_file, "r") as f:
            file_content = f.read()

        # Use Poscar.from_string() to parse the file content
        poscar = Poscar.from_string(file_content)
        structure = poscar.structure

        # Extract only molecule atoms
        molecule_sites = [site for site in structure.sites if site.species_string in molecule_atoms]

        if len(molecule_sites) < 2:
            raise ValueError("Molecule must have at least two atoms to compute distances.")

        distances = []

        if len(molecule_sites) == 2:
            # Biatomic case (e.g., N2, O2)
            distances.append(structure.lattice.get_distance_and_image(
                molecule_sites[0].frac_coords, molecule_sites[1].frac_coords
            )[0])

        elif len(molecule_sites) == 3:
            # Case: N2H (2N, 1H) → Find N-N and smallest N-H distance
            nitrogen_atoms = [site for site in molecule_sites if site.species_string == "N"]
            hydrogen_atoms = [site for site in molecule_sites if site.species_string == "H"]

            if len(nitrogen_atoms) == 2 and len(hydrogen_atoms) == 1:
                nn_distance = structure.lattice.get_distance_and_image(
                    nitrogen_atoms[0].frac_coords, nitrogen_atoms[1].frac_coords
                )[0]

                nh_distances = [
                    structure.lattice.get_distance_and_image(nitrogen.frac_coords, hydrogen_atoms[0].frac_coords)[0]
                    for nitrogen in nitrogen_atoms
                ]

                distances.append(nn_distance)
                distances.append(min(nh_distances))  # Choose the smallest N-H distance

        elif len(molecule_sites) == 4:
            # Case: NH3 (1N, 3H) → Compute 3 N-H distances and sort them
            nitrogen_atoms = [site for site in molecule_sites if site.species_string == "N"]
            hydrogen_atoms = [site for site in molecule_sites if site.species_string == "H"]

            if len(nitrogen_atoms) == 1 and len(hydrogen_atoms) == 3:
                nh_distances = [
                    structure.lattice.get_distance_and_image(nitrogen_atoms[0].frac_coords, h.frac_coords)[0]
                    for h in hydrogen_atoms
                ]

                distances.extend(sorted(nh_distances))  # Sort from smallest to largest

        return distances

    except Exception as e:
        print(f"Error processing CONTCAR file: {e}")
        return None
```

```python
from pymatgen.core import Structure
from pymatgen.core.periodic_table import Element
from pymatgen.io.vasp import Potcar
import numpy as np

def get_elneg(atom_type):
    """Returns the electronegativity of an atom type."""
    try:
        return Element(atom_type).X
    except ValueError:
        return None  # Return None if the element is invalid

def read_acf(acf_file):
    """Reads the ACF.dat file and extracts atomic positions and Bader charges."""
    surface_atoms = []
    with open(acf_file, 'r') as f:
        lines = f.readlines()

    for i, line in enumerate(lines[2:], start=1):  # Skip header lines, start indexing from 1
        parts = line.split()
        if len(parts) >= 5:  # Ensure enough data
            try:
                x, y, z, charge = map(float, parts[1:5])
                surface_atoms.append({'index': i - 1, 'coords': np.array([x, y, z]), 'bader_charge': charge})
            except ValueError:
                continue  # Skip if parsing fails
    
    return surface_atoms

def get_valence_electrons_from_potcar_pymatgen(potcar_file):
    """Reads a POTCAR file using pymatgen and extracts the number of valence electrons."""
    try:
        potcar = Potcar.from_file(potcar_file)
        return {entry.element: entry.nelectrons for entry in potcar}
    except FileNotFoundError:
        raise FileNotFoundError(f"The file {potcar_file} does not exist.")
    except Exception as e:
        raise RuntimeError(f"An error occurred while parsing the POTCAR file: {e}")

def get_element_types_from_poscar(poscar_file):
    """Reads the POSCAR file to get the ordered list of element types."""
    structure = Structure.from_file(poscar_file)
    return [site.species_string for site in structure.sites]  # Ordered list of atomic elements

def find_closest_surface_atom_old(poscar_file, acf_file, potcar_file, atom_types):
    """
    Finds the closest surface atom to each molecule atom while considering periodic boundary conditions.
    
    :param poscar_file: Path to the POSCAR file
    :param acf_file: Path to the ACF.dat file
    :param potcar_file: Path to the POTCAR file
    :param atom_types: List of atom types in the molecule (e.g., ['N', 'H'])
    :return: List of dictionaries containing closest surface atom data
    """
    # Load structure from POSCAR
    structure = Structure.from_file(poscar_file)
    
    # Extract molecule atoms
    molecule_atoms = [site for site in structure.sites if site.species_string in atom_types]
    
    # Load surface atom positions and their Bader charges
    surface_atoms = read_acf(acf_file)
    
    # Get correct element types from POSCAR (indexed by atomic positions)
    element_types = get_element_types_from_poscar(poscar_file)
    
    # Load ZVAL values from POTCAR
    zvals = get_valence_electrons_from_potcar_pymatgen(potcar_file)
    
    results = []
    
    for mol_atom in molecule_atoms:
        min_distance = float('inf')
        closest_surface_atom = None
        
        # Convert molecule atom coordinates to fractional
        mol_atom_frac = structure.lattice.get_fractional_coords(mol_atom.coords)

        # Find closest surface atom using combined distance methods
        for surf_atom in surface_atoms:
            surf_atom_frac = structure.lattice.get_fractional_coords(surf_atom['coords'])

            # **1. Compute direct Euclidean distance**
            direct_distance = np.linalg.norm(mol_atom.coords - surf_atom['coords'])

            # **2. Compute periodic image distance**
            periodic_distance, image = structure.lattice.get_distance_and_image(mol_atom_frac, surf_atom_frac)

            # **3. Ensure periodic boundary handling**
            wrapped_surf_coords = structure.lattice.get_cartesian_coords(surf_atom_frac + image)

            # **4. Compute corrected periodic distance**
            corrected_periodic_distance = np.linalg.norm(mol_atom.coords - wrapped_surf_coords)

            # **5. Choose the shortest valid distance**
            actual_distance = min(direct_distance, periodic_distance, corrected_periodic_distance)

            if actual_distance < min_distance:
                min_distance = actual_distance
                closest_surface_atom = surf_atom
        
        # Ensure we found a closest surface atom
        if closest_surface_atom:
            # Get correct surface atom type from POSCAR indexing
            surface_atom_index = closest_surface_atom['index']  # Already adjusted to 0-based indexing
            surface_element = element_types[surface_atom_index]  # Get element type

            # Get ZVAL for the surface atom
            zval = zvals.get(surface_element, None)  # Get ZVAL from POTCAR dictionary
            
            # Compute corrected Bader charge
            corrected_bader_charge = zval - closest_surface_atom['bader_charge'] if zval is not None else None
            
            # Get electronegativity safely
            elneg = get_elneg(surface_element) if surface_element else None

            # Store results
            results.append({
                'molecule_atom': mol_atom.species_string,
                'molecule_coords': mol_atom.coords,
                'closest_surface_atom_type': surface_element,
                'closest_surface_atom_index': surface_atom_index,  # Include the index from POSCAR
                'closest_surface_atom_coords': closest_surface_atom['coords'],
                'closest_surface_atom_bader_charge': closest_surface_atom['bader_charge'],
                'corrected_bader_charge': corrected_bader_charge,
                'closest_surface_atom_elneg': elneg,
                'distance': min_distance
            })
    
    return results
```

```python
#This function only finds THE closest atom to each atom in the final adsorbate in OUTCAR
from pymatgen.core import Structure
from pymatgen.core.periodic_table import Element
from pymatgen.io.vasp import Potcar
import numpy as np

def get_elneg(atom_type):
    """Returns the electronegativity of an atom type."""
    try:
        return Element(atom_type).X
    except ValueError:
        return None  # Return None if the element is invalid

def read_acf(acf_file):
    """Reads the ACF.dat file and extracts atomic positions and Bader charges."""
    surface_atoms = []
    with open(acf_file, 'r') as f:
        lines = f.readlines()

    for i, line in enumerate(lines[2:], start=1):  # Skip header lines, start indexing from 1
        parts = line.split()
        if len(parts) >= 5:  # Ensure enough data
            try:
                x, y, z, charge = map(float, parts[1:5])
                surface_atoms.append({'index': i - 1, 'coords': np.array([x, y, z]), 'bader_charge': charge})
            except ValueError:
                continue  # Skip if parsing fails
    
    return surface_atoms

def get_valence_electrons_from_potcar_pymatgen(potcar_file):
    """Reads a POTCAR file using pymatgen and extracts the number of valence electrons."""
    try:
        potcar = Potcar.from_file(potcar_file)
        return {entry.element: entry.nelectrons for entry in potcar}
    except FileNotFoundError:
        raise FileNotFoundError(f"The file {potcar_file} does not exist.")
    except Exception as e:
        raise RuntimeError(f"An error occurred while parsing the POTCAR file: {e}")

def get_element_types_from_poscar(poscar_file):
    """Reads the POSCAR file to get the ordered list of element types."""
    structure = Structure.from_file(poscar_file)
    return [site.species_string for site in structure.sites]  # Ordered list of atomic elements

def find_closest_surface_atom(poscar_file, acf_file, potcar_file, atom_types):
    """
    Finds the closest surface atom to each molecule atom while considering periodic boundary conditions.
    
    :param poscar_file: Path to the POSCAR file
    :param acf_file: Path to the ACF.dat file
    :param potcar_file: Path to the POTCAR file
    :param atom_types: List of atom types in the molecule (e.g., ['N', 'H'])
    :return: List of dictionaries containing closest surface atom data
    """
    # Load structure from POSCAR
    structure = Structure.from_file(poscar_file)
    
    # Extract molecule atoms
    molecule_atoms = [site for site in structure.sites if site.species_string in atom_types]
    
    # Load surface atom positions and their Bader charges
    surface_atoms = read_acf(acf_file)
    
    # Get correct element types from POSCAR (indexed by atomic positions)
    element_types = get_element_types_from_poscar(poscar_file)
    
    # Load ZVAL values from POTCAR
    zvals = get_valence_electrons_from_potcar_pymatgen(potcar_file)
    
    results = []
    
    for mol_atom in molecule_atoms:
        min_distance = float('inf')
        closest_surface_atom = None
        closest_surface_index = None  # Store POSCAR index

        # Convert molecule atom coordinates to fractional
        mol_atom_frac = structure.lattice.get_fractional_coords(mol_atom.coords)

        # Find closest surface atom using combined distance methods
        for surf_atom in surface_atoms:
            surface_index = surf_atom['index']  # Index from ACF file

            # Ensure we use POSCAR coordinates instead of ACF coordinates
            poscar_coords = structure.sites[surface_index].coords

            # **1. Compute direct Euclidean distance**
            direct_distance = np.linalg.norm(mol_atom.coords - poscar_coords)

            # **2. Compute periodic image distance**
            surf_atom_frac = structure.lattice.get_fractional_coords(poscar_coords)
            periodic_distance, image = structure.lattice.get_distance_and_image(mol_atom_frac, surf_atom_frac)

            # **3. Ensure periodic boundary handling**
            wrapped_surf_coords = structure.lattice.get_cartesian_coords(surf_atom_frac + image)

            # **4. Compute corrected periodic distance**
            corrected_periodic_distance = np.linalg.norm(mol_atom.coords - wrapped_surf_coords)

            # **5. Choose the shortest valid distance**
            actual_distance = min(direct_distance, periodic_distance, corrected_periodic_distance)

            if actual_distance < min_distance:
                min_distance = actual_distance
                closest_surface_atom = surf_atom
                closest_surface_index = surface_index

        # Ensure we found a closest surface atom
        if closest_surface_atom:
            # Get correct surface atom type from POSCAR indexing
            surface_element = element_types[closest_surface_index]  # Get element type

            # Get ZVAL for the surface atom
            zval = zvals.get(surface_element, None)  # Get ZVAL from POTCAR dictionary
            
            # Compute corrected Bader charge
            corrected_bader_charge = zval - closest_surface_atom['bader_charge'] if zval is not None else None
            
            # Get electronegativity safely
            elneg = get_elneg(surface_element) if surface_element else None

            # Store results
            results.append({
                'molecule_atom': mol_atom.species_string,
                'molecule_coords': mol_atom.coords,
                'closest_surface_atom_type': surface_element,
                'closest_surface_atom_index': closest_surface_index,  # Include the index from POSCAR
                'closest_surface_atom_coords': structure.sites[closest_surface_index].coords,  # Use POSCAR coordinates
                'closest_surface_atom_bader_charge': closest_surface_atom['bader_charge'],
                'corrected_bader_charge': corrected_bader_charge,
                'closest_surface_atom_elneg': elneg,
                'distance': min_distance
            })
    
    return results
```

```python
#I want to now add another feature to dataset by adding orbital band centers
from pymatgen.io.vasp import Vasprun
from pymatgen.electronic_structure.core import OrbitalType
import numpy as np
from scipy.integrate import trapezoid  # <-- Safe alternative

def calculate_orbital_band_info(vasprun_path, target_indices=None, emin=-10, emax=10):
    vasprun = Vasprun(vasprun_path)
    cdos = vasprun.complete_dos
    energies = cdos.energies - cdos.efermi

    mask = (energies >= emin) & (energies <= emax)
    energies_masked = energies[mask]

    results = {
        "selected_indices": target_indices if target_indices else [],
    }

    orbitals = {
        "s": OrbitalType.s,
        "p": OrbitalType.p,
        "d": OrbitalType.d,
    }

    def compute_band_stats(energies, total_dos, weighted_sum):
        band_sum = float(trapezoid(total_dos, energies))
        band_center = float(trapezoid(weighted_sum, energies) / band_sum) if band_sum != 0 else None
        return band_center, band_sum

    for label, orbital in orbitals.items():
        # Selected atoms
        total_dos = np.zeros_like(energies)
        weighted_sum = np.zeros_like(energies)
        valid = False

        if target_indices:
            for i in target_indices:
                site = cdos.structure[i]
                spd_dos = cdos.get_site_spd_dos(site)
                if orbital not in spd_dos:
                    continue
                dos = spd_dos[orbital].get_densities()
                total_dos += dos
                weighted_sum += dos * energies
                valid = True

        total_dos_masked = total_dos[mask]
        weighted_sum_masked = weighted_sum[mask]
        if valid:
            center, summation = compute_band_stats(energies_masked, total_dos_masked, weighted_sum_masked)
            results[f"selected_{label}band_center"] = center
            results[f"selected_{label}band_sum"] = summation
        else:
            results[f"selected_{label}band_center"] = None
            results[f"selected_{label}band_sum"] = None

        # Total atoms
        total_total_dos = np.zeros_like(energies)
        total_weighted_sum = np.zeros_like(energies)

        for i, site in enumerate(cdos.structure):
            spd_dos = cdos.get_site_spd_dos(site)
            if orbital not in spd_dos:
                continue
            dos = spd_dos[orbital].get_densities()
            total_total_dos += dos
            total_weighted_sum += dos * energies

        total_total_dos = total_total_dos[mask]
        total_weighted_sum = total_weighted_sum[mask]

        if np.any(total_total_dos):
            total_center, total_sum = compute_band_stats(energies_masked, total_total_dos, total_weighted_sum)
            results[f"total_{label}band_center"] = total_center
            results[f"total_{label}band_sum"] = total_sum
        else:
            results[f"total_{label}band_center"] = None
            results[f"total_{label}band_sum"] = None

    return results
```
